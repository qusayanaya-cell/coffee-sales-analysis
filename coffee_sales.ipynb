```python
import pandas as pd
data_frame = pd.read_csv("archive\Coffe_sales.csv")
```

```python
df1 = pd.DataFrame(data_frame,columns=['Time_of_Day','money','coffee_name','Weekday','cash_type'])
df1
```

```python
import seaborn as sns

sns.set()

graph = sns.scatterplot(x='Weekday',y='money',data = df1)
graph
```

```python
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import HoverTool

output_notebook()

TOOLTIPS = HoverTool(tooltips=[
    ('(cash_type)', '(@cash_type)'),
    ('(coffee_name)', '(@coffee_name)'),
])

p = figure(title="Coffee sales",
           x_range=['Fri', 'Sat', 'Sun', 'Mon', 'Tue', 'Wed', 'Thu'],
           x_axis_label='Weekday', y_axis_label='money',
           width=700, height=700, tools=[TOOLTIPS])

p.scatter('Weekday', 'money', size=10, source=df1)
show(p)
```
